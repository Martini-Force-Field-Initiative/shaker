# Mapping complex systems with `Shaker`

This tutorial demonstrates a simple workflow for mapping a complex atomistic membrane containing multiple components, such as this DOPC:Cholesterol lipid bilayer.


SHAKER uses a simple dictionary to describe how atomistic atoms are grouped
into CG beads. This mapping dictionary is reused across tools, including
trajectory mapping, SASA analysis, topology generation, and visualization.

The mapping dictionary can be conveniently generated using
[cgbuilder](https://lp0lp.github.io/cgbuilder/), which allows you to
interactively group AA atoms into CG beads and export the resulting mapping in
the **SHAKER format**.

```python
mapping = {
    ## Resname
    "DOPC": {
    # Bead name;  Mapping.
        "NC3": {"atoms": ['N', 'C13', 'H13A', 'H13B', 'H13C', 'C14', 'H14A', 'H14B', 'H14C', 'C15', 'H15A','H15B', 'H15C', 'C12', 'H12A', 'H12B', 'C11', 'H11A', 'H11B'] },     
        "PO4": {"atoms": ['P', 'O13', 'O14', 'O11', 'O12'] },
        "GL1": {"atoms": ['C2', 'HS', 'O21', 'C21', 'O22', 'C1', 'HA', 'HB' ] },
        ...
    },
    
    ## Resname
    "CHL1": {
    # Bead name;  Mapping.
        "ROH": {"atoms": ['C2', 'C2', 'C4', 'C4', 'O3', 'O3', 'O3', 'O3'] },      
        "R1":  {"atoms": ['C6', 'C6', 'C6', 'C6','C6', 'C5', 'C7','H6','H7A','H7B'] },     
        ...
    },
}
```

The mapping dictionary with one top-level entry per residue name, where each entry defines a CG bead:

* `"MOL"` – residue name in the atomistic system
* `"R1", "R2"` – CG bead names
* `"atoms"` – atom names mapped to that bead
* `"type"` *(optional)* – Martini bead type
* `"charge"` *(optional)* – bead charge

Bead coordinates are computed as the **center of geometry** of the listed atoms.
Atoms can appear multiple times to apply simple weighting:

```python
"atoms": ['C2', 'C2', 'C4', 'C4', 'O3', 'O3', 'O3', 'O3'] 
```

Once the mapping is defined, `shaker.mapper.map_aa2cg` can be used to generate a
**CG-mapped trajectory** from the reference AA simulation.
